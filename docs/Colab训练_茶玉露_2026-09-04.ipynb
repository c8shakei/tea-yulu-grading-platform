```python
# ===== 恢复脚本：不重训，直接抢救已训练权重并同步到 Drive =====
from pathlib import Path
import json, shutil
import yaml
import matplotlib.pyplot as plt
import pandas as pd
from datetime import datetime
from ultralytics import YOLO
from google.colab import drive

drive.mount('/content/drive', force_remount=False)
DRIVE_SAVE = Path("/content/drive/MyDrive/BISHE/models")
DRIVE_SAVE.mkdir(parents=True, exist_ok=True)

# 生成 Colab 版 data.yaml（绝对路径）
data_yaml = Path("/content/datasets/tea_yulu/data.yaml")
cfg = yaml.safe_load(data_yaml.read_text(encoding="utf-8"))
cfg["path"] = str(data_yaml.parent)
resolved = data_yaml.parent / "data_colab.yaml"
resolved.write_text(yaml.safe_dump(cfg, sort_keys=False, allow_unicode=True), encoding="utf-8")

# ===== 检测模型：重算 test 指标 + 曲线 + 同步 =====
detect_dir = Path("/content/runs/detect/tea_yulu_detect/yolov8n")
det_model = YOLO(str(detect_dir / "weights" / "best.pt"))
det_m = det_model.val(data=str(resolved), split="test")
test_metrics = {
    "mAP50-95": float(det_m.box.map),
    "mAP50": float(det_m.box.map50),
    "precision": float(det_m.box.mp),
    "recall": float(det_m.box.mr),
    "created_at": datetime.now().isoformat(),
}
(detect_dir / "test_metrics.json").write_text(json.dumps(test_metrics, indent=2, ensure_ascii=False), encoding="utf-8")
print("检测 test 指标:", test_metrics)

df = pd.read_csv(detect_dir / "results.csv")
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
for s in ["train/box_loss", "val/box_loss"]:
    if s in df: axes[0,0].plot(df["epoch"], df[s], label=s)
axes[0,0].set_title("Box Loss"); axes[0,0].legend()
for s in ["train/cls_loss", "val/cls_loss"]:
    if s in df: axes[0,1].plot(df["epoch"], df[s], label=s)
axes[0,1].set_title("Cls Loss"); axes[0,1].legend()
if "metrics/mAP50(B)" in df: axes[1,0].plot(df["epoch"], df["metrics/mAP50(B)"], label="mAP50")
if "metrics/mAP50-95(B)" in df: axes[1,0].plot(df["epoch"], df["metrics/mAP50-95(B)"], label="mAP50-95")
axes[1,0].set_title("mAP"); axes[1,0].legend()
if "metrics/precision(B)" in df: axes[1,1].plot(df["epoch"], df["metrics/precision(B)"], label="Precision")
if "metrics/recall(B)" in df: axes[1,1].plot(df["epoch"], df["metrics/recall(B)"], label="Recall")
axes[1,1].set_title("Precision/Recall"); axes[1,1].legend()
plt.tight_layout(); fig.savefig(detect_dir / "training_curves.png", dpi=300); plt.close(fig)

shutil.copy(detect_dir / "weights" / "best.pt", DRIVE_SAVE / "best_detect.pt")
for fn in ["test_metrics.json", "training_curves.png", "results.csv"]:
    s = detect_dir / fn
    if s.exists(): shutil.copy(s, DRIVE_SAVE / fn)

# ===== 分类模型：重算 val 指标 + 同步 =====
cls_dir = Path("/content/datasets/tea_yulu_cls")
cls_run = Path("/content/runs/classify/tea_yulu_cls/yolov8n-cls")
cls_model = YOLO(str(cls_run / "weights" / "best.pt"))
cls_m = cls_model.val(data=str(cls_dir))
val_metrics = {
    "top1_accuracy": float(cls_m.top1),
    "top5_accuracy": float(cls_m.top5),
    "created_at": datetime.now().isoformat(),
}
(cls_run / "val_metrics.json").write_text(json.dumps(val_metrics, indent=2, ensure_ascii=False), encoding="utf-8")
print("分类 val 指标:", val_metrics)

shutil.copy(cls_run / "weights" / "best.pt", DRIVE_SAVE / "best_cls.pt")
for fn in ["val_metrics.json", "results.csv"]:
    s = cls_run / fn
    if s.exists(): shutil.copy(s, DRIVE_SAVE / fn)

print("==== 全部同步完成 ====")
print([p.name for p in DRIVE_SAVE.iterdir()])
```
